```python
class Student:
    def __init__(self, name, age, gpa):
        self.name = name
        self.age = age
        self.gpa = gpa

    # Getter and setter for name
    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    # Getter and setter for age
    def get_age(self):
        return self.age

    def set_age(self, age):
        self.age = age

    # Getter and setter for GPA
    def get_gpa(self):
        return self.gpa

    def set_gpa(self, gpa):
        self.gpa = gpa
```

**Q1. Side by side —** list each getter/setter pair the assistant
wrote. Next to each, write what the equivalent line looks like using
*your* `Student` class from Part A (direct attribute access).
**Q1. Side by side —**

1. `get_name()` → `self.name` = `student.name`
2. `set_name(name)` → `self.name = name` = `student.name = name`
3. `get_age()` → `self.age` = `student.age`
4. `set_age(age)` → `self.age = age` = `student.age = age`
5. `get_gpa()` → `self.gpa` = `student.gpa`
6. `set_gpa(gpa)` → `self.gpa = gpa` = `student.gpa = gpa`

**Q2. Why not? —** Python doesn't generally need getters and setters
the way some other languages do. Why not — and if you genuinely needed
to intercept attribute access later (e.g. for validation), what
Python feature would you reach for instead?
Python allows direct access to attributes, so getters and setters are usually unnecessary. If I needed to control or validate an attribute later, I would use Python's @property feature, which allows me to keep simple attribute-style access while adding validation or other logic behind the scenes.

**Q3. Which would you maintain? —** in a class with 15 attributes
instead of 3, which style would you rather work in six months from now,
and why?
I would rather maintain the direct attribute access style. With 15 attributes, writing separate getters and setters for every attribute would create a lot of repetitive code and make the class harder to read and maintain. Direct access is simpler, and if I later need validation for a specific attribute, I can use @property for just that attribute.
